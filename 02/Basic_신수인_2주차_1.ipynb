```python
df.isna().sum() #True가 1이라 더해서 0보다 크면 결손 데이터 O
```

```python
df['Cabin'] = df['Cabin'].fillna('C000') #혹은 inplace = True
df.head()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna('S')
df.isna().sum()
```

```python
df['Name_len'] = df['Name'].apply(lambda x : len(x)) #람다를 쓰면 속도가 더 빠르고 복잡한 데이터 가공 가능
df[['Name', 'Name_len']].head()
```

```python
df['ChildAdult'] = df['Age'].apply(lambda x : 'Child' if x <= 15 else 'Adult' )
df[['Age', 'ChildAdult']].head()
```

```python
df['Age_cat'] = df['Age'].apply(lambda x: 'Child' if x <= 15 else ('Adult' if x <= 60 else 'Elderly'))

df['Age_cat'].value_counts()
```
