## DataFrame과 리스트, 딕셔너리, 넘파이 ndarray 상호 변환
사이킷런에서는 넘파이 ndarray를 입력 인자로 사용하는 경우가 대부분이다. 따라서 DataFrame과 넘파이 ndarray 상호 간의 변환은 중요하다.

#### 리스트, 딕셔너리, 넘파이 ndarray -> DataFrame

```python
# 1차원 리스트, ndarray -> DataFrame
import numpy as np
col_name1=['col1']
list1=[1,2,3]
array1=np.array(list1)
df_list1=pd.DataFrame(list1, columns=col_name1)
print('1차원 리스트로 만든 DataFrame:\n', df_list1)
df_array1=pd.DataFrame(array1, columns=col_name1)
print('1차원 ndarray로 만든 DataFrame:\n', df_array1)
```

DataFrame은 기본적으로 2차원 data이기 때문에 변환 대상인 data 또한 2차원 이하여야한다.

```python
# 2차원
list2=[[1,2,3], [11,12,13]]
array2=np.array(list2)
col_name2 = ['col1', 'col2', 'col3'] # 2X3 사이즈 행렬이기 때문에 3개의 칼럼명이 필요
df_list2=pd.DataFrame(list2, columns=col_name2)
df_array2=pd.DataFrame(array2, columns=col_name2)

print('2차원 리스트로 만든 DataFrame:\n', df_list2)
print('2차원 ndarray로 만든 DataFrame:\n', df_array2)
```

```python
# 딕셔너리는 key가 칼럼명으로 매핑, value가 리스트 형(또는 ndarray) 칼럼 데이터로 매핑
dict ={'col':[1,11], 'col2':[2,12], 'col3':[3,13]}
df_dict=pd.DataFrame(dict)
print('딕셔너리로 만든 DataFrame:\n', df_dict)
```

#### DataFrmae -> 리스트, 딕셔너리, 넘파이 ndarray
DataFrame -> ndarray 에서는 values가 사용됨

```python
# DataFrame을 ndarray로 변환
array3 = df_dict.values
print('df_dict.values 타입:', type(array3), 'df_dict.values shape', array3.shape)
print(array3)
```

```python
# DataFrame을 리스트로 변환
list3 = df_dict.values.tolist() # DataFrame -> ndarray -> list
print('df_dict.values.tolist()타입:', type(list3))
print(list3)

# DataFrame을 딕셔너리로 변환
dict3 = df_dict.to_dict('list') # to_dict()의 인자를 'list'로 쓰면, 딕셔너리의 value가 list 형태로 변환된다.
print('df_dict.to_dict() 타입:', type(dict3))
print(dict3)
```

#### apply lambda 식으로 데이터 가공
파이썬의 함수 선언하는 것을 한 줄로 간단하게 만든 식

```python
def get_square(a):
    return a**2
print('3의 제곱은:', get_square(3))
```

```python
lambda_square = lambda x : x ** 2
print('3의 제곱은:', lambda_square(3))
```

lambda x : x**2 <br>
: 왼쪽의 x는 입력인자, 오른쪽은 반환할 결과

```python
# 여러 인자를 넣고 싶으면 map()함수를 사용하면 된다.
a=[1,2,3]
squares=map(lambda x : x**2, a)
list(squares)
```

```python
# map()은 series에서 레코드별로 함수를 적용하려는 함수이다.
a=[1,2,3]
b=pd.DataFrame(a)
squares=b.map(lambda x : x**2)
squares
```

apply()는 DataFrame이나 series에 사용자 정의 함수를 적용하려는 함수이다.

```python
titanic_df['Name_len']=titanic_df['Name'].apply(lambda x:len(x))
titanic_df[['Name', 'Name_len']].head(3)
```

```python
titanic_df['Child_Adult']=titanic_df['Age'].apply(lambda x:'Child' if x<=15 else 'Adult') # 반환값을 if절보다 먼저 기술해야한다.
titanic_df[['Age', 'Child_Adult']].head(8)
```

if, else만 지원하고 if, else if, else는 지원하지 않는다. <br>
else if를 구현하려면 else에 ()를 내포해서 if else를 기술해야한다.

```python
titanic_df['Age_cat']=titanic_df['Age'].apply(lambda x:'Child'if x<=15 else('Adult'if x<=60 else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```

분류할 케이스가 많으면 별도의 함수를 만들어서 적용하는 게 나을 수도 있다.

```python
# 나이에 따라 세분화된 분류를 수행하는 함수 생성
def get_category(age):
    cat=''
    if age <= 5: cat='Baby'
    elif age <=12: cat='Child'
    elif age <=18: cat='Teenager'
    elif age <=25: cat='Student'
    elif age <=35: cat='Young Adult'
    elif age <=60: cat='Adult'
    else : cat='Elderly'
   
    return cat

titanic_df['Age_cat']=titanic_df['Age'].apply(lambda x : get_category(x))
titanic_df[['Age', 'Age_cat']].head()
```
