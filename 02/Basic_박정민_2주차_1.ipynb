## 04_2 : DataFrame과 리스트, 딕셔너리, 넘파이 ndarray 상호 변환

### 넘파이 ndarray, 리스트, 딕셔너리를 DataFrame으로 변환하기
- DataFrame은 리스트,넘파이 ndarray와 달리 칼럼명 O
- 변환 시 칼럼명 지정(지정하지 않으면 자동으로 할당)
- 2차원 이하의 데이터들만 DataFrame으로 변환 가능

1. 1차원 형태의 리스트와 ndarray DataFrame으로 변환하기

```python
#1차원 데이터이기 때문에 칼럼명 1개 필요함
```

```python
import numpy as np

col_name1=['col1']
list1 = [1, 2, 3]
array1 = np.array(list1)
print('array1 shape:', array1.shape )
# 리스트를 이용해 DataFrame 생성.
df_list1 = pd.DataFrame(list1, columns=col_name1)
print('1차원 리스트로 만든 DataFrame:\n', df_list1)
# 넘파이 ndarray를 이용해 DataFrame 생성.
df_array1 = pd.DataFrame(array1, columns=col_name1)
print('1차원 ndarray로 만든 DataFrame:\n', df_array1)
```

2. 2차원 형태의 데이터 DataFrame으로 변환하기

```python
# 3개의 칼럼명이 필요함.
col_name2=['col1', 'col2', 'col3']

# 2행x3열 형태의 리스트와 ndarray 생성한 뒤 이를 DataFrame으로 변환.
list2 = [[1, 2, 3],
         [11, 12, 13]]
array2 = np.array(list2)
print('array2 shape:', array2.shape )
df_list2 = pd.DataFrame(list2, columns=col_name2)
print('2차원 리스트로 만든 DataFrame:\n', df_list2)
df_array2 = pd.DataFrame(array2, columns=col_name2)
print('2차원 ndarray로 만든 DataFrame:\n', df_array2)
```

3. 딕셔너리를 DataFrame으로 변환하기

```python
# Key는 컬럼명으로 매핑, Value는 리스트 형(또는 ndarray)
dict = {'col1':[1, 11], 'col2':[2, 22], 'col3':[3, 33]}
df_dict = pd.DataFrame(dict)
print('딕셔너리로 만든 DataFrame:\n', df_dict)
```

### DataFrame을 넘파이 ndarray, 리스트, 딕셔너리로 변환하기

```python
# values 메소드 활용
array3 = df_dict.values
print('df_dict.values 타입:', type(array3), 'df_dict.values shape:', array3.shape)
print(array3)
```

## 04_6 : 정렬, Aggregation 함수, GroupBy 적용

- DataFrame, Series의 정렬 - sort_values()
- 주요 파라미터 : by, ascending, inplace
- by : 입력한 칼럼을 기준으로 정렬 수행
- ascending : True면 오름차순(기본), False면 내림차순
- inplace : True면 정렬 결과 원데이터 반영, False면 원데이터는 그대로 두고 정렬한 것을 따로 반환

```python
# 'Name'칼럼을 기준으로 오름차순 정렬
titanic_sorted = titanic_df.sort_values(by=['Name'])
titanic_sorted.head(3)
```

- groupby() 이용하기
- 입력 파라미터 by에 칼럼 입력하면 대상 칼럼으로 groupby 됨

```python
# 위 df에 aggregation 함수를 호출하면 groupby() 대상 칼럼을 제외한 모든 칼럼에 적용됨(SQL과 다른부분)
titanic_groupby = titanic_df.groupby('Pclass').count()
titanic_groupby
```

```python
# 특정 칼럼만 aggregation 함수를 적용하려며 Groupby()로 반환된 DataFrameGroupBy 객체에 해당 칼럼을 필터링한 뒤 적용
# SQL의 경우 대상 칼럼을 모두 Select 절에 나열
# 'PassengerId', 'Survived'칼럼만 필터링하여 적용
titanic_groupby = titanic_df.groupby('Pclass')[['PassengerId', 'Survived']].count()
titanic_groupby
```

```python
# 서로 다른 aggregation 함수 적용할 경우
# SQL : 그냥 나열
# DataFrame groupby() : agg() 내에 입력하여 사용
titanic_df.groupby('Pclass')['Age'].agg([max, min])
```

## 04_7 : 결손 데이터 처리하기  
- 결손 데이터 : 칼럼에 값이 없는 NULL 상태, NaN으로 표시
- 머신러닝 알고리즘에서의 NaN : 기본적으로 처리 X -> 다른값으로 대체하지 않으면 연산 시 제외가 됨

- isna()로 결손 데이터 여부 확인

```python
titanic_df.isna().head(3)
```

```python
# sum() 추가하여 결손 데이터 개수 구하기  
titanic_df.isna( ).sum( )
```

- fillna()로 Missing 데이터 대체하기
- fillna()로 반환 값을 다시 받거나, inplace=True 파라미터를 추가하여야 실제 데이터 세트 값이 반영된다.

```python
# titanic_df의 'Cabin'칼럼의 NaN값을 'C000'으로 대체  
titanic_df['Cabin'] = titanic_df['Cabin'].fillna('C000')
titanic_df.head(3)
```

```python
# 'Age' 칼럼의 NaN 값을 평균 나이로, 'Embarked'칼럼의 NaN 값을 'S'로 대체하여 모든 결손 데이터 처리 
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())
titanic_df['Embarked'] = titanic_df['Embarked'].fillna('S')
titanic_df.isna().sum()
```

## 04_8 : apply lambda 식으로 데이터 가공
- Pandas에서는 apply 함수에 lambda 식을 결합하여 DataFrame이나 Series의 레코드별로 데이터 가공 가능

```python
# get_square(a) 함수가 있다고 가정
def get_square(a):
    return a**2

print('3의 제곱은:',get_square(3))
```

```python
# 위와 같은 함수의 선언과 함수 내의 처리를 한 줄 식으로 쉽게 변환하는 게 lambda함수
lambda_square = lambda x : x ** 2
print('3의 제곱은:',lambda_square(3))
```

```python
# 여러 개의 값을 입력 인자로 써야 할 때 : map() 함수 사용
a=[1,2,3]
squares = map(lambda x : x**2, a)
list(squares)
```

- Pandas DataFrame 에서의 lambda 식도 동일한 방식으로 적용

```python
# 'Name' 칼럼의 문자열 개수를 별도의 칼럼인 'Name_len'에 생성
titanic_df['Name_len']= titanic_df['Name'].apply(lambda x : len(x))
titanic_df[['Name','Name_len']].head(3)
```

```python
# if else 절을 이용하여 'Age'가 15 미만이면 'Child', 아니면 'Adult'로 구분하는 새로운 칼럼 생성
titanic_df['Child_Adult'] = titanic_df['Age'].apply(lambda x : 'Child' if x <=15 else 'Adult' )
titanic_df[['Age','Child_Adult']].head(8)
```

- lambda식에서 if else 절 쓸 때 주의할 점
1. if 식보다 반환 값 먼저 서술(':' 오른쪽에 반환 값이 존재해야 하기 때문)
2. else if 는 지원하지 않는다

```python
# else 절을 ()로 내포하여 () 내에서 다시 if else를 적용하여 else if를 사용하는 방법 예시 
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : 'Child' if x<=15 else ('Adult' if x <= 60 else 
                                                                                  'Elderly'))
titanic_df['Age_cat'].value_counts()
```

```python
# else 절을 계속 내포하여 사용하는 게 부담스러울 때 : 아예 별도의 함수를 만든다.
# 5세 이하 : Baby, 12세 이하 : Child, 25세 이하 : Student, 35세 이하 : Young Adult, 60세 이하 : Adult, 그 이상 : Elderly 로 분류하는 함수 생성
def get_category(age):
    cat = ''
    if age <= 5: cat = 'Baby'
    elif age <= 12: cat = 'Child'
    elif age <= 18: cat = 'Teenager'
    elif age <= 25: cat = 'Student'
    elif age <= 35: cat = 'Young Adult'
    elif age <= 60: cat = 'Adult'
    else : cat = 'Elderly'
    
    return cat

# lambda 식에 위에서 생성한 get_category( ) 함수를 반환값으로 지정. 
# get_category(X)는 입력값으로 ‘Age’ 컬럼 값을 받아서 해당하는 cat 반환
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x))
titanic_df[['Age','Age_cat']].head()
```
