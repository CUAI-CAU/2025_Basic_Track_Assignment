Guide of what can go inside [] after DataFrame object.
- `[]` after DataFrame object is different from `[]` after Series object or `[]` after numpy array.
- only column name/list, slices, boolean index can go inside `[]` after DataFrame object.
- while slicing is possible but not recommended.

- you don't need to use loc or iloc when you want to select all rows in some/all column(s).
- loc vs iloc
    - loc: name(label) based indexing
    - iloc: integer based indexing
- when using loc, unlike iloc, it includes the end point.

```python
titanic_df = pd.read_csv('/mnt/c/Users/USER/Downloads/titanic/train.csv')
```

```python
titanic_df['Name_len'] = titanic_df.Name.apply(lambda x: len(x))
titanic_df[['Name', 'Name_len']].head(3)
```

```python
titanic_df['Child_Adult'] = titanic_df['Age'].apply(lambda x: 'Child' if x <= 15 else 'Adult')
titanic_df[['Age', 'Child_Adult']].head(8)
```

```python
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x: 'Child' if x <= 15 else ('Adult' if x <= 60 else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```

```python
def get_category(age):
    if age <= 5:
        cat = 'Baby'
    elif age <= 12:
        cat = 'Child'
    elif age <= 18:
        cat = 'Teenager'
    elif age <= 25:
        cat = 'Student'
    elif age <= 35:
        cat = 'Young Adult'
    elif age <= 60:
        cat = 'Adult'
    else: cat = 'Elderly'
    
    return cat

titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x: get_category(x))
titanic_df[['Age', 'Age_cat']].head()
```
