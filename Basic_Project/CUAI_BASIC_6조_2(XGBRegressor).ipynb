### 전처리  

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np
import pandas as pd

def preprocess(df):
    # 1. ID 컬럼 삭제
    if 'ID' in df.columns:
        df = df.drop(columns=['ID'])

    # 2. '분야' 결측치 처리
    if '분야' in df.columns:
        df['분야'] = df['분야'].fillna('Unknown')

    # 3. '직원 수', '고객수(백만명)' median으로 결측치 대체
    for col in ['직원 수', '고객수(백만명)']:
        if col in df.columns:
            median_val = df[col].median()
            df[col] = df[col].fillna(median_val)

    # 4. '기업가치(백억원)' 문자열 전처리
    def convert_to_lower_limit(value):
        if pd.isna(value):
            return np.nan
        if isinstance(value, str):
            if value == '6000이상':
                return 6000.0
            if '-' in value:
                return float(value.split('-')[0])
            return float(value)
        return value

    if '기업가치(백억원)' in df.columns:
        df['기업가치(백억원)'] = df['기업가치(백억원)'].apply(convert_to_lower_limit)
        df['기업가치(백억원)'] = df['기업가치(백억원)'].fillna(df['기업가치(백억원)'].median())

    # 5. Label Encoding ('인수여부', '상장여부')
    for col in ['인수여부', '상장여부']:
        if col in df.columns:
            le = LabelEncoder()
            df[col] = le.fit_transform(df[col])

    # 6. One-hot Encoding ('국가', '분야', '투자단계')
    onehot_cols = ['국가', '분야', '투자단계']
    df = pd.get_dummies(df, columns=[col for col in onehot_cols if col in df.columns])

    return df
```

```python
# 1. 'ID' 칼럼 삭제
df.drop(columns=['ID'], inplace=True)
```

```python
# 2. '분야' 칼럼 결측치 'Unknown' 처리
df['분야'] = df['분야'].fillna('Unknown')
```

```python
# 3. '직원 수', '고객수(백만명)' 칼럼 결측치 'median' 대체
for col in ['직원 수', '고객수(백만명)']:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)
```

```python
# 4. '기업가치' 칼럼 전처리

# 범주형 칼럼 '기업가치'를 하한값으로 변환하는 함수
def convert_to_lower_limit(value):

    # 값이 NaN이면 그대로 반환
    if pd.isna(value):
        return np.nan

    # 값이 '문자열'이면 특수 문자 '-' 포함 여부 확인
    if isinstance(value, str):

        # 값이 '6000이상' -> 6000.0으로 변환
        if value == '6000이상':
            return 6000.0

        # 값이 특수 문자 '-' 포함한 범주형 -> 하한값('-' 앞의 값)만 추출
        if '-' in value:
            return float(value.split('-')[0])

        # 값이 '-'없이 숫자만 문자열 형태로 있는 경우 -> float 변환
        return float(value)

    return value

# 함수 적용
df['기업가치(백억원)'] = df['기업가치(백억원)'].apply(convert_to_lower_limit)

# 남은 결측치(NaN) -> 전체의 중앙값(median)으로 대체
df['기업가치(백억원)'] = df['기업가치(백억원)'].fillna(df['기업가치(백억원)'].median())
```

```python
# 5. 범주형 데이터 처리
from sklearn.preprocessing import LabelEncoder

# '인수여부', '상장여부' -> Label Encoding
label_cols = ['인수여부', '상장여부']
for col in label_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

# '국가', '분야', '투자단계' -> One - Hot Encoding
onehot_cols = ['국가', '분야', '투자단계']
df = pd.get_dummies(df, columns=onehot_cols)
```

```python
# 이상치 확인
outlier_summary = []

# 추가
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    outlier_summary.append({
        "Feature": col,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "IQR": round(IQR, 2),
        "LowerBound": round(lower, 2),
        "UpperBound": round(upper, 2),
        "OutlierCount": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df.sort_values(by="OutlierCount", ascending=False)
```

```python
# 기준 연도 설정
CURRENT_YEAR = 2025

# 업력 파생변수 생성 (한글 컬럼명 기준) (업력: 쉽게 말해 기업의 나이)
df['업력'] = CURRENT_YEAR - df['설립연도'] + 1
```

```python
# 작은 수 ε 설정 (0으로 나누는 것 방지)
epsilon = 1e-5
```

```python
# 매출대비투자비율 생성
epsilon = 1e-5
df['매출대비투자비율'] = df['연매출(억원)'] / (df['총 투자금(억원)'] + epsilon)

# 성공확률 기준 그룹 나누기
threshold = 0.7
high = df[df['성공확률'] > threshold]['매출대비투자비율']
low = df[df['성공확률'] <= threshold]['매출대비투자비율']

# Mann-Whitney U test
from scipy.stats import mannwhitneyu
stat, p = mannwhitneyu(high, low, alternative='two-sided')

print(f"U 통계량: {stat:.2f}")
print(f"p-value: {p:.5f}")

if p < 0.05:
    print("✅ 통계적으로 유의한 차이가 있음 (파생변수 의미 있음)")
else:
    print("⚠️ 통계적으로 유의하지 않음 (파생변수 의미 약함)")
```

```python
# 기업가치 성장 속도 파생변수 생성
epsilon = 1e-5
df['기업가치성장속도'] = df['기업가치(백억원)'] / (df['업력'] + epsilon)

# 고성공 vs 저성공 그룹 분리
threshold = 0.7
high_group = df[df['성공확률'] > threshold]['기업가치성장속도']
low_group = df[df['성공확률'] <= threshold]['기업가치성장속도']

# Mann-Whitney U test
stat, p = mannwhitneyu(high_group, low_group, alternative='two-sided')

# 결과 출력
print(f"U 통계량: {stat:.2f}")
print(f"p-value: {p:.5f}")

if p < 0.05:
    print("✅ 통계적으로 유의한 차이가 있음 (파생변수 의미 있음)")
else:
    print("⚠️ 통계적으로 유의하지 않음 (파생변수 의미 약함)")
```

```python
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
from collections import Counter
import numpy as np
```

```python
# Weighted MAE 평가 함수
def weighted_mae(y_test, y_preds, weights):
    y_test = np.asarray(y_test)
    y_preds = np.asarray(y_preds)
    weights = np.asarray(weights)

    return np.sum(weights * np.abs(y_test - y_preds)) / np.sum(weights)
```

```python
# 가중치 계산 함수
def compute_weights(y_test):
    y_test = np.asarray(y_test)
    freq = Counter(y_test)
    weights = np.array([1.0 / freq[val] for val in y_test])
    return weights
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split

# 전처리
df_train = pd.read_csv(r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/train.csv')
df_train = preprocess(df_train)

df_test = pd.read_csv(r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/test.csv')
df_test = preprocess(df_test)

# 학습 데이터 분리
X_features = df_train.drop(columns='성공확률')
y_label = df_train['성공확률']
X_train, X_valid, y_train, y_valid = train_test_split(X_features, y_label, test_size=0.2, random_state=42)

# DMatrix로 변환
dtrain = xgb.DMatrix(X_train, label=y_train)
dvalid = xgb.DMatrix(X_valid, label=y_valid)
dtest = xgb.DMatrix(df_test[X_features.columns])
dval_eval = xgb.DMatrix(X_valid)

# 하이퍼파라미터 설정
params = {
    'objective': 'reg:squarederror',
    'eval_metric': 'mae',
    'learning_rate': 0.01,
    'max_depth': 6
}

# 모델 학습
model = xgb.train(
    params=params,
    dtrain=dtrain,
    num_boost_round=5000,
    evals=[(dtrain, 'train'), (dvalid, 'valid')],
    early_stopping_rounds=200,
    verbose_eval=10
)

# 검증 예측 및 평가
w_preds = model.predict(dval_eval)
weights = compute_weights(y_valid)
score = weighted_mae(y_valid, w_preds, weights)
print(f"Weighted MAE: {score:.4f}")

# 제출용 예측 및 저장
final_predictions = model.predict(dtest)

sample_submission = pd.read_csv(r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/sample_submission.csv')
sample_submission['성공확률'] = final_predictions
sample_submission.to_csv('./baseline_submission.csv', index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
```

```python
# 학습 및 평가 함수
def train_and_eval(X_train, y_train, X_val, y_val, X_final, filename):
    model = XGBRegressor(n_estimators=5000, learning_rate=0.01, max_depth=6, eval_metric='mae')
    model.fit(X_train, y_train)

    val_pred = model.predict(X_val)
    weights = compute_weights(y_val)
    score = weighted_mae(y_val, val_pred, weights)
    print(f"{filename} → Weighted MAE: {score:.4f}")

    final_pred = model.predict(X_final)
    sub = pd.read_csv(sub_path)
    sub['성공확률'] = final_pred
    sub.to_csv(f'./{filename}_submission.csv', index=False, encoding='utf-8-sig')
```

```python
# 데이터 경로
train_path = r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/train.csv'
test_path = r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/test.csv'
sub_path = r'C:/Users/csh34/Documents/jupyter_notebook/basic_dacon/sample_submission.csv'

# 데이터 로딩
df_train = pd.read_csv(train_path)
df_test = pd.read_csv(test_path)

df_train = preprocess(df_train)
df_test = preprocess(df_test)

# 업력 계산용 기준년도 설정
CURRENT_YEAR = 2023
epsilon = 1e-5

# 업력 파생변수 생성
df_train['업력'] = CURRENT_YEAR - df_train['설립연도'] + 1
df_test['업력'] = CURRENT_YEAR - df_test['설립연도'] + 1

# 파생 변수 생성
df_train['매출대비투자비율'] = df_train['연매출(억원)'] / (df_train['총 투자금(억원)'] + epsilon)
df_test['매출대비투자비율'] = df_test['연매출(억원)'] / (df_test['총 투자금(억원)'] + epsilon)

df_train['기업가치성장속도'] = df_train['기업가치(백억원)'] / (df_train['업력'] + epsilon)
df_test['기업가치성장속도'] = df_test['기업가치(백억원)'] / (df_test['업력'] + epsilon)

# 타겟
y = df_train['성공확률']

# 케이스별 X 구성
X0 = df_train.drop(columns=['성공확률', '매출대비투자비율', '기업가치성장속도'], errors='ignore') # 파생 없음
X1 = df_train[['매출대비투자비율']] # 매출대비투자비율만
X2 = df_train[['기업가치성장속도']] # 기업가치성장속도만
X3 = df_train[['매출대비투자비율', '기업가치성장속도']] # 둘 다 사용

# 테스트셋 구성
X0_test = df_test[X0.columns]
X1_test = df_test[X1.columns]
X2_test = df_test[X2.columns]
X3_test = df_test[X3.columns]

# 훈련/검증 분리
X0_train, X0_val, y_train, y_val = train_test_split(X0, y, test_size=0.2, random_state=42)
X1_train, X1_val = X1.loc[X0_train.index], X1.loc[X0_val.index]
X2_train, X2_val = X2.loc[X0_train.index], X2.loc[X0_val.index]
X3_train, X3_val = X3.loc[X0_train.index], X3.loc[X0_val.index]

# 케이스별 실행
train_and_eval(X0_train, y_train, X0_val, y_val, X0_test, 'baseline_파생0')
train_and_eval(X1_train, y_train, X1_val, y_val, X1_test, 'baseline_매출투자비율만')
train_and_eval(X2_train, y_train, X2_val, y_val, X2_test, 'baseline_기업가치성장속도만')
train_and_eval(X3_train, y_train, X3_val, y_val, X3_test, 'baseline_파생2개')
```
