### ndarray를 편리하게 생성하기 - arange,zeros,ones

특정 크기,차원을 가진 ndarray를 연속값/0/1로 초기화해 쉽게 생성해야 할 경우
arrange(),zeros(),ones()로 쉽게 ndarray를 생성

테스트용으로 데이터 만들거나 대규모 데이터 일괄적 초기화 할 경우

```python
sequence_array=np.arange(10)
print(sequence_array)
print(sequence_array.dtype,sequence_array.shape)
```

zeros() : 함수 인자로 튜플형태의 shape값을 입력 시 모든 값을 0으로 채운 shape을 가진 ndarray로 반환

```python
zero_array=np.zeros((3,2),dtype='int32')
print(zero_array)
print(zero_array.dtype,zero_array.shape)
```

ones() : 함수 인자로부터 튜플 형태의 shape값 입력 시 모든 값을 1로 채운 해당 shape를 가진 ndarray를 반환

함수인자로 dtype설정x-> float64형의 데이터로 ndarray를 채움

```python
one_array=np.ones((3,2))
print(one_array)
print(one_array.dtype,one_array.shape)
```

### ndarray의 차원과 크기를 변경하는 reshape()

ndarray를 특정 차원,크기로 변환
변환하는 크기를 함수 인자로 부여함

```python
array1=np.arange(10)
print('array1:\n',array1)
```

```python
array2=array1.reshape(2,5)
print('array2:\n',array2)
```

```python
array3=array1.reshape(5,2)
print('array3:\n',array3)
```

```python
array1.reshape(4,3)
```

지정된 사이즈로 변경 불가 시 오류 발생

-1을 인자로 사용 시 ndarray와 호환되는 새로운 shape로 변환해줌

```python
array1=np.arange(10)
print(array1)
```

```python
array2=array1.reshape(-1,5)
print('array2 shape:',array2.shape)
```

고정된 5개 칼럼에 맞는 로우를 자동으로 새롭게 생성해 변환
-> 10개의 1차원 데이터와 호환될 수 있는 고정된 5개 칼럼에 맞는 로우 개수는 2이므로 2*5의 2차원ndarray로 변환

```python
array3=array1.reshape(5,-1)
print('array3 shape:',array3.shape)
```

고정된 5개의 로우에 맞는 컬럼은 2이므로 5*2의 2차원 ndarray로 변환

```python
array1=np.arange(10)
array4=array1.reshape(-1,4)
```

10개의 1차원 데이터를 고정된 4개의 칼럼가진 로우로는 변경 x

reshape(-1,1)같은걸로 자주 사용
원본 ndarray가 어떤 형태라도 2차원,여러개의 로우를 가지되 반드시 1개 칼럼 가진 ndarray로 변환됨을 보장

```python
#1차원 ndarray를 2차원 ndarray로 변환
array6=array1.reshape(-1,1)
print('array6:\n',array6.tolist())
```

```python
print('array6 shape:',array6.shape)
```

### 팬시 인덱싱

리스트나 ndarray로 인덱스 집합을 지정하면 해당 위치의 인덱스에 해당하는 ndarray를 반환하는 인덱싱 방식

```python
array1d=np.arange(start=1,stop=10)
array2d=array1d.reshape(3,3)
```

```python
array3d=array2d[[0,1],2]
print('array2d[[0,1],2] ->',array3d.tolist())
```

로우축에 팬시인덱싱인 [0,1]을 칼럼축에 단일값 인덱싱2를 적용 -> (0,2),(1,2)로 적용되어 [3,6]반환

```python
array4d=array2d[[0,1],0:2]
print('array2d[[0,1],0:2]->',array4d.tolist())
```

((0,0),(0,1)),(1,0),(1,1))인덱싱 적용 -> [[1,2],[4,5]]반환

```python
array5=array2d[[0,1]]
print('array2d[[0,1]]->',array5.tolist())
```

((0,:),(1,:))인덱싱 적용 -> [[1,2,3],[4,5,6]]반환

### 행렬의 정렬 = sort()와 argsort()

np.sort()는 원 행렬 유지한채 원행렬의 정렬된 행렬을 반환
ndarray.sort()는 원행렬 자체를 정렬한 형태로 변환하며 반환값은 None

```python
org_array=np.array([3,1,9,5])
print('원본행렬:',org_array)

sort_array1=np.sort(org_array)
print('np.sort()호출 후 반환된 정렬 행렬:',sort_array1)
```

```python
print('np.sort()호출 후 원본 행렬:',org_array)
```

```python
sort_array2=org_array.sort()
print('org_array.sort()호출 후 반환된 행렬:',sort_array2)
print('org_array.sort()호출 후 원본 행렬:',org_array)
```

```python
sort_array1_desc=np.sort(org_array)[::-1]
print('내림차순으로 정렬:',sort_array1_desc)
```

- 행렬이 2차원 이상일 경우 axis축값 설정으로 로우/칼럼방향으로 정렬 수행 가능

```python
array2d=np.array([[8,12],[7,1]])
sort_array2d_axis0=np.sort(array2d,axis=0)
print('로우방향으로 정렬:\n',sort_array2d_axis0)

sort_array2d_axis1=np.sort(array2d,axis=1)
print('칼럼방향으로 정렬:\n;',sort_array2d_axis1)
```

### 정렬된 행렬의 인덱스를 반환하기

원본행렬이 정렬되었을 때 기존 원본 행렬의 원소에 대한 인덱스를 필요할 때 np.argsort() 사용
-> 정렬 행렬의 원본 행렬 인덱스를 nparray형으로 반환

```python
org_array=np.array([3,1,9,5])
sort_indices=np.argsort(org_array)
print(type(sort_indices))
print('행렬정렬시 원본 행렬의 인덱스',sort_indices)
```

```python
org_array=np.array([3,1,9,5])
sort_indices_desc=np.argsort(org_array)[::-1]
print('행렬내림차순 정렬 시 원본 행렬의 인덱스:',sort_indices_desc)
```

- 넘파이의 데이터 추출에서 많이 사용됨

```python
import numpy as np
name_array=np.array(['john','mike','sarah','kate','samuel'])
score_array=np.array([78,95,84,98,88])

sort_indices_asc=np.argsort(score_array)
print('성적오름차순 정렬 시 score_array의 인덱스:',sort_indices_asc)
print('성적옮차순으로 naem_array이름출력:',name_array[sort_indices_asc])
```

### 선형대수 연산 - 행렬 내적과 전치 행렬 구하기

- 행렬 내적(행렬곱)

```python
A=np.array([[1,2,3],
            [4,5,6]])
B=np.array([[7,8],
            [9,10],
            [11,12]])
```

```python
dot_product=np.dot(A,B)
print('행렬내적 결과:\n',dot_product)
```

- 전치행렬

원행렬에서 행과 열의 위치를 교환한 원소로 구성한 행렬

행렬 A의 전치행렬 -> A^T

```python
A=np.array([[1,2],
            [3,4]])
transpose_mat=np.transpose(A)
print('A의 전치행렬:\n',transpose_mat)
```
