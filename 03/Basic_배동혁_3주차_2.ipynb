```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import KFold
import numpy as np

iris=load_iris()
features=iris.data
label=iris.target
dt_clf=DecisionTreeClassifier(random_state=156)

kfold=KFold(n_splits=5)
cv_accuracy=[]
print('붗꽃 데이터 세트 크기:', features.shape[0])
```

```python
# 반복 횟수를 세기 위한 변수 : n_iter
n_iter=0

# kfold.split(features) : feature 데이터를 k개의 fold로 나눠서, train_index와 test_index를 만든다. (index로 출력)
# 여기서 label은 split으로 나누지 않았는데 그 이유는 label은 numpy 데이터라서 train_index 대로 따라가면 되는 것이다.
# 예를 들어, features = np.array([[1], [2], [3], [4], [5]]), label = np.array([0, 0, 1, 1, 1]) 이라고 치면,
# Train: [1 2 3 4] Test: [0] 이렇게 나누어 지고, label은 이 index를 그대로 따라서 가는 것이다.

for train_index, test_index in kfold.split(features):
    X_train, X_test=features[train_index], features[test_index]
    y_train, y_test=label[train_index], label[test_index]

    dt_clf.fit(X_train, y_train)
    pred=dt_clf.predict(X_test)
    n_iter+=1

    accuracy=np.round(accuracy_score(y_test,pred),4)
    train_size=X_train.shape[0]
    test_size=X_test.shape[0]
    print('\n #{0} 교차 검증 정확도 "{1}, 학습 데이터 크기 : {2}, 검증 데이터 크기 : {3}'.format(n_iter,accuracy,train_size, test_size))
    print('\n #{0} 검증 세트 인덱스:{1}'.format(n_iter,test_index))
    cv_accuracy.append(accuracy)

print('\n 평균 검증 정확도: ',np.mean(cv_accuracy))
```

Stratified K-Fold : 특정 label 값이 특이하게 많거나 적은 data에서 클래스 비율을 원본 데이터와 최대한 비슷하가ㅔ 유지.

```python
import pandas as pd

iris=load_iris()
#iris_df : iris.data로 dataframe 생성, 각 열들의 제목은 iris.feature_names(sepal length 등등)
iris_df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
# iris_df에 label열을 추가, 값은 iris.target
iris_df['label']=iris.target
iris_df['label'].value_counts()
```

```python
kfold=KFold(n_splits=3)
n_iter=0

for train_index, test_index in kfold.split(iris_df):
    n_iter+=1
    # label_train = iris_df의 label열의 train_index의 데이터값 (0,1,2중 하나)
    # label_test = iris_df의 label열의 test_index의 데이터값 (0,1,2중 하나)
    label_train=iris_df['label'].iloc[train_index]
    label_test=iris_df['label'].iloc[test_index]
    print('## 교차 검증: {0}'.format(n_iter))
    print('학습 레이블 데이터 분포:\n', label_train.value_counts())
    print('검증 레이블 데이터 분포:\n', label_test.value_counts())
```

```python
from sklearn.model_selection import StratifiedKFold

skf=StratifiedKFold(n_splits=3)
n_iter=0

for train_index, test_index in skf.split(iris_df, iris_df['label']):
    n_iter+=1
    label_train=iris_df['label'].iloc[train_index]
    label_test=iris_df['label'].iloc[test_index]
    print('## 교차 검증: {0}'.format(n_iter))
    print('학습 레이블 데이터 분포:\n', label_train.value_counts())
    print('검증 레이블 데이터 분포:\n', label_test.value_counts())
```

```python
dt_clf=DecisionTreeClassifier(random_state=156)

skfold=StratifiedKFold(n_splits=3)
n_iter=0
cv_accuracy=[]

# StratifiedKFold의 split() 호출 시 반드시 레이블 데이터 세트도 추가 입력 필요 
for train_index, test_index in skfold.split(features,label):
    X_train, X_test=features[train_index], features[test_index]
    y_train, y_test=label[train_index], label[test_index]

    dt_clf.fit(X_train, y_train)
    pred=dt_clf.predict(X_test)

    n_iter+=1
    accuracy=np.round(accuracy_score(y_test,pred),4)
    train_size=X_train.shape[0]
    test_size=X_test.shape[0]
    print('#{0} 교차 검증 정확도:{1}, 학습 데이터 크기 : {2}, 검증 데이터 크기 : {3}'.format(n_iter,accuracy,train_size,test_size))
    print('#{0} 검증 세트 인덱스:{1}'.format(n_iter,test_index))
    cv_accuracy.append(accuracy)

print('## 교차 검증별 정확도:',np.round(cv_accuracy,4))
print('## 평균 검증 정확도 :', np.round(np.mean(cv_accuracy),4))
```

Hyperparameter : 기계학습 모델을 학습시키기 전에 사용자가 직접 설정해야 하는 값들. 모델의 구조나 학습 방식에 영향을 주며, 학습 과정 중에 자동으로 조정되지 않고 고정된 값으로 사용된다.
학습률(Learning rate): 모델이 얼마나 빠르게 또는 천천히 학습할지를 결정함.
에폭(Epoch): 전체 데이터를 몇 번 반복해서 학습할지를 정함.
배치 크기(Batch size): 한 번에 학습에 사용하는 데이터의 개수.
은닉층 수와 각 층의 뉴런 수: 인공신경망의 구조를 정의하는 값들.
정규화 계수(L2 정규화 등): 과적합을 방지하기 위해 사용하는 규제의 세기.
드롭아웃 비율(Dropout rate): 일부 뉴런을 임의로 끄는 비율로, 일반화 성능 향상에 도움됨.

Hyperparameter는 모델의 성능에 큰 영향을 미치므로, Hyperparamete 튜닝 과정을 통해 최적의 값을 찾아야 한다. 
이때는 그리드 서치(Grid Search), 랜덤 서치(Random Search), 베이지안 최적화(Bayesian Optimization) 등의 기법을 사용할 수 있다.

GridSearchCV :  사이킷런에서 제공하는 하이퍼파라미터 튜닝 도구. 사용자가 지정한 여러 하이퍼파라미터 조합을 체계적으로 탐색하여 최적의 조합을 찾는다.
1. 사용자가 지정한 하이퍼파라미터 후보값들을 조합한다.
2. 각 조합에 대해 Cross Validation을 수행한다.
3. 검증 성능이 가장 좋은 최적의 파라미터 조합과 모델을 반환한다.

```python
grid_parameters = {'max_depth' : [1,2,3],
                   'min_samples_split':[2,3]}
# 이렇게 되면 max_depth=1, min_samples_split=1 / max_depth=1, min_samples_split=2 ....... 이렇게 6번 계산해서 제일 좋은 것을 알려준다.
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

iris_data=load_iris()
X_train, X_test, y_train, y_test=train_test_split(iris_data.data,iris_data.target,test_size=0.2,random_state=121)

dtree=DecisionTreeClassifier()

parameters={'max_depth':[1,2,3],'min_samples_split':[2,3]}

import pandas as pd

# GridSearchCV (decisiontreeclassifier 모델 사용, max_depth와 min_samples_split 하이퍼파라미터 조합, 3분할 교차검증)
grid_dtree=GridSearchCV(dtree,param_grid=parameters, cv=3, refit=True)

grid_dtree.fit(X_train,y_train)

# cv_results_: GridSearchCV가 내부적으로 계산한 모든 실험 결과를 담고 있는 딕셔너리.
scores_df=pd.DataFrame(grid_dtree.cv_results_)
scores_df[['params','mean_test_score','rank_test_score','split0_test_score','split1_test_score','split2_test_score']]
```

```python
print('gridSearchCV 최적 파라미터:', grid_dtree.best_params_)
print('GridSearchCV 최고 정확도:{0:.4f}'.format(grid_dtree.best_score_))
```

```python
estimator=grid_dtree.best_estimator_

pred=estimator.predict(X_test)
print('테스트 데이터 세트 정확도: {0:.4f}'.format(accuracy_score(y_test,pred)))
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# 학습 데이터는 0부터 10까지, 테스트 데이터는 0부터 5까지 값을 가지는 데이터 세트로 생성
# Scaler 클래스의 fit(), transform()은 2차원 이상 데이터만 가능하므로 reshape(-1,1)로 차원 변경
train_array=np.arange(0,11).reshape(-1,1)
test_array=np.arange(0,6).reshape(-1,1)

# MinMaxScaler 객체에 별도의 feature_range 파라미터 값을 지정하지 않으면 0~1의 값으로 변환.
scaler=MinMaxScaler()

# fit() 하게 되면 train_array 데이터의 최솟값이 0, 최댓값이 10으로 설정
scaler.fit(train_array)

# 1/10 scale로 train_array 데이터 변환함. 원본 10->1로 변환됨.
train_scaled=scaler.transform(train_array)

print('원본 train_array 데이터:', np.round(train_array.reshape(-1),2))
print('scale된 train_array 데이터:', np.round(train_scaled.reshape(-1),2))
```

```python
# fit() 하게 되면 test_array 데이터의 최솟값이 0, 최댓값이 5로 설정
scaler.fit(test_array)

# 1/5 scale로 test_array 데이터 변환함. 원본 5->1로 변환됨.
test_scaled=scaler.transform(test_array)

print('원본 test_array 데이터:', np.round(test_array.reshape(-1),2))
print('scale된 test_array 데이터:', np.round(test_scaled.reshape(-1),2))
```
