### [붓꽃 품종 예측하기 모델 구현]
#### 1. 사이킷런에서 사용할 모듈을 import

```python
#load_iris()를 이용하여 데이터 세트를 형성하여 load
from sklearn.datasets import load_iris

#DecisionTreeClassifier 적용하여 의사결정트리 알고리즘 구현
from sklearn.tree import DecisionTreeClassifier

#train_test_split() 함수 적용하여 데이터 세트를 학습 데이터와 테스트 데이터로 분류
from sklearn.model_selection import train_test_split
```

#### 2. 구성된 데이터 세트를 df으로 변환하여 피처들, 데이터 값 -> 어떻게 구성돼 있는 지 확인

```python
import pandas as pd

# 붓꽃 데이터 세트를 로딩합니다. 
iris = load_iris()

# 데이터프레임으로 변환하기 위한 사전 작업
# iris.data(피처 데이터들), iris.target(레이블 데이터), iris.feature_names(피처칼럼의 이름), iris.target_names(레이블 이름)등은 
# scikit-learn에서 load_iris()함수를 썼을 때 반환되도록 미리 정의해 둔 구성요소이고, 언제든 꺼내 쓸 수 있다.

# iris.data(피처 데이터)를 변수에 저장 / 특징 데이터  
iris_data = iris.data

# iris.target(레이블 데이터)을 변수에 저장 / 정답 데이터  
iris_label = iris.target
print('iris target값:', iris_label)
# 레이블 명 print
print('iris target명:', iris.target_names)

# 붓꽃 데이터 세트를 자세히 보기 위해 DataFrame으로 변환.
# feature 칼럼 생성
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)
# label 칼럼 생성
iris_df['label'] = iris.target
iris_df.head(3)
```

- 피처 : sepel length, sepal width, petal length, petal width
- 레이블 : 0(Setosa), 1(versicolor), 2(virginica)

#### 3. 학습용 데이터와 테스트용 데이터 분리
- train_test_split()으로 분리 : test_size 파라미터 입력값의 비율로 분리 
- 전체 데이터에서 테스트용 데이터를 분리해 내는 것은 반드시 필요한 작업

```python
# test_size 파라미터 값을 0.2로 설정하면 테스트 데이터 20%, 학습 데이터 80%로 분할된다.
# X : 피처 데이터(train/test), y : 레이블 데이터(train/test)
# random_state : 할 때마다 동일한 결과 나오게 함
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label, 
                                                    test_size=0.2, random_state=11)
```

#### 4. 의사결정트리로 학습 및 예측 수행

```python
# DecisionTreeClassifier 객체 생성 
dt_clf = DecisionTreeClassifier(random_state=11)

# 학습 수행 
# fit() 메서드 사용
dt_clf.fit(X_train, y_train)
```

```python
# 학습이 완료된 DecisionTreeClassifier 객체에서 테스트 데이터 세트로 예측 수행.
# predict() 메서드 사용
pred = dt_clf.predict(X_test)
```

#### 5. 예측 성능 평가

```python
# 정확도 측정(예측 결과가 실제 레이블 값과 얼마나 정확하게 맞는지)
# accuracy_score() 함수 사용
# 이때 앞서 사용하지 않은 y_test(레이블 테스트 값)을 사용하는 것. 
from sklearn.metrics import accuracy_score
print('예측 정확도: {0:.4f}'.format(accuracy_score(y_test,pred)))
```

## 1. 학습 데이터/테스트 데이터 셋 분리 – train_test_split()

#### 1) train_test_split() 함수로 학습 데이터/테스트 데이터 분리하기

- random_state : 지정해 주면 호출할 때마다 동일한 학습/테스트용 데이터 받을 수 있음

#### 2) 테스트 데이터 전체의 30%로 설정, random_state 파라미터는 121로 변경

```python
# 분류 모델 - 의사결정나무 모델
from sklearn.tree import DecisionTreeClassifier
# 평가 지표 - 정확도
from sklearn.metrics import accuracy_score
# 데이터셋 로딩
from sklearn.datasets import load_iris
# 학습데이터/테스트데이터 분리  
from sklearn.model_selection import train_test_split

dt_clf = DecisionTreeClassifier( )
iris_data = load_iris()

# 파라미터 1 : 피처 데이터 세트 / 파라미터 2 : 레이블 데이터 세트
# test_size 파라미터 : 테스트 데이터 크기 설정 / random_state 파라미터 : 호출할 때마다 동일한 학습/테스트용 세트 생성하게
X_train, X_test, y_train, y_test = train_test_split(iris_data.data, iris_data.target, 
                                                    test_size=0.3, random_state=121)
```

## 2. 교차 검증
- 데이터 편중을 막기 위해 별도의 여러 세트로 구성된 학습 데이터 세트와 검증 데이터 세트에서 학습/평가를 수행

p) 모델이 해당 테스트 데이터에만 과적합되어 다른 테스트용 데이터에 적용할 경우 성능이 떨어짐

S) 교차검증한다(과적합 방지)

#### 1. K 폴드 교차 검증
- 가장 보편적인 교차 검증 방법
- K개의 데이터 폴드 세트를 만들어 K번의 학습과 검증 평가 수행 -> 평균 내기
- KFlod 클래스 / StratifiedKFold 클래스 존재

##### 1) KFlod 클래스  

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import KFold
import numpy as np

iris = load_iris()
features = iris.data
label = iris.target
dt_clf = DecisionTreeClassifier(random_state=156)

# 5개의 폴드 세트로 분리하는 KFold 객체와 폴드 세트별 정확도를 담을 리스트 객체 생성.
kfold = KFold(n_splits=5)
cv_accuracy = []
print('붓꽃 데이터 세트 크기:',features.shape[0])
```

```python
n_iter = 0

# KFold객체의 split( ) 호출하면 폴드 별 학습용, 검증용 테스트의 로우 인덱스를 array로 반환  
for train_index, test_index  in kfold.split(features):
    # kfold.split( )으로 반환된 인덱스를 이용하여 학습용, 검증용 테스트 데이터 추출
    X_train, X_test = features[train_index], features[test_index]
    y_train, y_test = label[train_index], label[test_index]
    #학습 및 예측 
    dt_clf.fit(X_train , y_train)    
    pred = dt_clf.predict(X_test)
    n_iter += 1
    # 반복 시 마다 정확도 측정 
    accuracy = np.round(accuracy_score(y_test,pred), 4)
    train_size = X_train.shape[0]
    test_size = X_test.shape[0]
    print('\n#{0} 교차 검증 정확도 :{1}, 학습 데이터 크기: {2}, 검증 데이터 크기: {3}'
          .format(n_iter, accuracy, train_size, test_size))
    print('#{0} 검증 세트 인덱스:{1}'.format(n_iter,test_index))
    cv_accuracy.append(accuracy)
    
# 개별 iteration별 정확도를 합하여 평균 정확도 계산 
print('\n## 평균 검증 정확도:', np.mean(cv_accuracy)) 
```

##### 2. Stratified K 폴드
- 불균형 분포도의 레이블 데이터 집합인 경우 사용
- KFold는 원본 데이터 집합의 레이블 분포를 학습 및 테스트 세트에 제대로 분배 못하는 경우가 있음.(작지만 중요한 데이터의 피처들이 제대로 반영 안되는 경우 등)

```python
import pandas as pd

#붓꽃 데이터 세트 데이터 프레임으로 형성
iris = load_iris()

iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['label']=iris.target
iris_df['label'].value_counts()
```

```python
kfold = KFold(n_splits=3)
# kfold.split(X)는 폴드 세트를 3번 반복할 때마다 달라지는 학습/테스트 용 데이터 로우 인덱스 번호 반환. 
n_iter =0
for train_index, test_index  in kfold.split(iris_df):
    n_iter += 1
    label_train= iris_df['label'].iloc[train_index]
    label_test= iris_df['label'].iloc[test_index]
    print('## 교차 검증: {0}'.format(n_iter))
    print('학습 레이블 데이터 분포:\n', label_train.value_counts())
    print('검증 레이블 데이터 분포:\n', label_test.value_counts())
    
```

```python
# 동일한 분석을 StratifiedKFlod로 수행
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=3)
n_iter=0

for train_index, test_index in skf.split(iris_df, iris_df['label']):
    n_iter += 1
    label_train= iris_df['label'].iloc[train_index]
    label_test= iris_df['label'].iloc[test_index]
    print('## 교차 검증: {0}'.format(n_iter))
    print('학습 레이블 데이터 분포:\n', label_train.value_counts())
    print('검증 레이블 데이터 분포:\n', label_test.value_counts())
```

```python
# 입력 age에 따라 구분값을 반환하는 함수 설정. DataFrame의 apply lambda식에 사용. 
def get_category(age):
    cat = ''
    if age <= -1: cat = 'Unknown'
    elif age <= 5: cat = 'Baby'
    elif age <= 12: cat = 'Child'
    elif age <= 18: cat = 'Teenager'
    elif age <= 25: cat = 'Student'
    elif age <= 35: cat = 'Young Adult'
    elif age <= 60: cat = 'Adult'
    else : cat = 'Elderly'
    
    return cat

# 막대그래프의 크기 figure를 더 크게 설정 
plt.figure(figsize=(10,6))

#X축의 값을 순차적으로 표시하기 위한 설정 
group_names = ['Unknown', 'Baby', 'Child', 'Teenager', 'Student', 'Young Adult', 'Adult', 'Elderly']

# lambda 식에 위에서 생성한 get_category( ) 함수를 반환값으로 지정. 
# get_category(X)는 입력값으로 'Age' 컬럼값을 받아서 해당하는 cat 반환
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x))
sns.barplot(x='Age_cat', y = 'Survived', hue='Sex', data=titanic_df, order=group_names)
titanic_df.drop('Age_cat', axis=1, inplace=True)
plt.show()
```

```python
from sklearn import preprocessing

def encode_features(dataDF):
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = preprocessing.LabelEncoder()
        le = le.fit(dataDF[feature])
        dataDF[feature] = le.transform(dataDF[feature])
        
    return dataDF

titanic_df = encode_features(titanic_df)
titanic_df.head()
```

```python
# 남은 문자열 카테고리 피처 숫자형 카테고리 피처로 변환
# 레이블 인코딩
from sklearn.preprocessing import LabelEncoder

# Null 처리 함수
def fillna(df):
    df['Age'].fillna(df['Age'].mean(), inplace=True)
    df['Cabin'].fillna('N', inplace=True)
    df['Embarked'].fillna('N', inplace=True)
    df['Fare'].fillna(0, inplace=True)
    return df

# 머신러닝 알고리즘에 불필요한 피처 제거
def drop_features(df):
    df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
    return df

# 레이블 인코딩 수행.
def format_features(df):
    df['Cabin'] = df['Cabin'].str[:1]
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = LabelEncoder()
        le = le.fit(df[feature])
        df[feature] = le.transform(df[feature])
    return df

# 앞에서 설정한 데이터 전처리 함수 호출
def transform_features(df):
    df = fillna(df)
    df = drop_features(df)
    df = format_features(df)
    return df
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 결정트리, Random Forest, 로지스틱 회귀를 위한 사이킷런 Classifier 클래스 생성
dt_clf = DecisionTreeClassifier(random_state=11)
rf_clf = RandomForestClassifier(random_state=11)
lr_clf = LogisticRegression(solver='liblinear')

# DecisionTreeClassifier 학습/예측/평가
dt_clf.fit(X_train , y_train)
dt_pred = dt_clf.predict(X_test)
print('DecisionTreeClassifier 정확도: {0:.4f}'.format(accuracy_score(y_test, dt_pred)))

# RandomForestClassifier 학습/예측/평가
rf_clf.fit(X_train , y_train)
rf_pred = rf_clf.predict(X_test)
print('RandomForestClassifier 정확도:{0:.4f}'.format(accuracy_score(y_test, rf_pred)))

# LogisticRegression 학습/예측/평가
lr_clf.fit(X_train , y_train)
lr_pred = lr_clf.predict(X_test)
print('LogisticRegression 정확도: {0:.4f}'.format(accuracy_score(y_test, lr_pred)))
```

```python
from sklearn.model_selection import KFold

def exec_kfold(clf, folds=5):
    # 폴드 세트를 5개인 KFold객체를 생성, 폴드 수만큼 예측결과 저장을 위한  리스트 객체 생성.
    kfold = KFold(n_splits=folds)
    scores = []
    
    # KFold 교차 검증 수행. 
    for iter_count , (train_index, test_index) in enumerate(kfold.split(X_titanic_df)):
        # X_titanic_df 데이터에서 교차 검증별로 학습과 검증 데이터를 가리키는 index 생성
        X_train, X_test = X_titanic_df.values[train_index], X_titanic_df.values[test_index]
        y_train, y_test = y_titanic_df.values[train_index], y_titanic_df.values[test_index]
        
        # Classifier 학습, 예측, 정확도 계산 
        clf.fit(X_train, y_train) 
        predictions = clf.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)
        scores.append(accuracy)
        print("교차 검증 {0} 정확도: {1:.4f}".format(iter_count, accuracy))     
    
    # 5개 fold에서의 평균 정확도 계산. 
    mean_score = np.mean(scores)
    print("평균 정확도: {0:.4f}".format(mean_score)) 
# exec_kfold 호출
exec_kfold(dt_clf , folds=5) 
```

```python
from sklearn.model_selection import GridSearchCV

parameters = {'max_depth':[2,3,5,10],
             'min_samples_split':[2,3,5], 'min_samples_leaf':[1,5,8]}

grid_dclf = GridSearchCV(dt_clf , param_grid=parameters , scoring='accuracy' , cv=5)
grid_dclf.fit(X_train , y_train)

print('GridSearchCV 최적 하이퍼 파라미터 :',grid_dclf.best_params_)
print('GridSearchCV 최고 정확도: {0:.4f}'.format(grid_dclf.best_score_))
best_dclf = grid_dclf.best_estimator_

# GridSearchCV의 최적 하이퍼 파라미터로 학습된 Estimator로 예측 및 평가 수행. 
dpredictions = best_dclf.predict(X_test)
accuracy = accuracy_score(y_test , dpredictions)
print('테스트 세트에서의 DecisionTreeClassifier 정확도 : {0:.4f}'.format(accuracy))
```
