```python
# ML모델 학습을 위해 fit(), 학습된 모델의 예측을 위해 predict()
# 지도학습의 경우위 두 개를 사용하여 사이킷런 클래스는 간단하게 학습과 예측 결과를 반환할 수 있음
# 분류 알고리즘을 구현한 클래스 = classifire, 회귀 알고리즘은 regression, 이 두 개를 합쳐서 esimator 클래스라고 부름(내부에서 fit, predict 구현하고 있음)
```

```python
#cross_val_score()과 같은 evaluation 함수, GridSearchCV 처럼 하이퍼 파라미터 튜닝을 지원하는 클래스는 estimator를 인자로 받는다
#그리고 이 인자를 통해서 함수 내에서 estimator의 fit과 predict를 호출하여 각각 평가를 하고 하이퍼 파라미터 튜닝을 함
```

```python
#비지도학습인 차원 축소, 클러스터링, 피처 추츨도 fit(), transform()을 대부분 적용한다. 
#이때의 fit은 학습을 의미 x, 입력데이터에 맞추어 데이터를 변환하기 위한 사전구조 맞춤 작업. 이후 실제 작업은 transform()을 통해 진행
#fit_transform()도 존재하나 주의가 필요함. 
```
