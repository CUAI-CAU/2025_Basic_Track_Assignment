```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestRegressor
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# UCI Boston Housing 데이터셋 URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/housing/housing.data"

# 컬럼 이름 지정 (원래 데이터셋에 헤더가 없어서 직접 추가해야 함)
column_names = [
    "CRIM", "ZN", "INDUS", "CHAS", "NOX", "RM", "AGE", "DIS", "RAD", 
    "TAX", "PTRATIO", "B", "LSTAT", "MEDV"
]

# 데이터 로드
bostonDF = pd.read_csv(url, delim_whitespace=True, names=column_names)

# Feature & Target 분리
y_target = bostonDF['MEDV']  # 주택 가격
X_data = bostonDF.drop(['MEDV'], axis=1, inplace = False)

rf = RandomForestRegressor(random_state = 0, n_estimators = 1000)
neg_mse_scores = cross_val_score(rf, X_data, y_target, scoring = 'neg_mean_squared_error', cv = 5)
rmse_scores = np.sqrt(-1*neg_mse_scores)
avg_rmse = np.mean(rmse_scores)

print(' 5 겨차 검증의 개별 Negative MSE scores: ', np.round(neg_mse_scores, 2))
print(' 5 교차 검증의 개별 RMSE socres : ', np.round(rmse_scores, 2))
print(' 5 교차 검증의 평균 RMSE : {0:.3f} '.format(avg_rmse))
```
